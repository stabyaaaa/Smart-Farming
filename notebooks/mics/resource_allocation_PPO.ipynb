The choice of scaling methods is based on the distribution characteristics of each feature. StandardScaler is used for temperature because it follows a normal distribution, ensuring a mean of 0 and a standard deviation of 1. RobustScaler is applied to nitrogen (N), phosphorus (P), and potassium (K) since they have multiple peaks and potential outliers, making it more robust against extreme values. PowerTransformer (Yeo-Johnson) is used for humidity and yield because both are right-skewed, and this transformation helps normalize their distribution. MinMaxScaler is applied to pH as it appears discrete and bounded between a fixed range, making normalization between 0 and 1 a suitable choice. If pH is categorical, one-hot encoding would be better to preserve its categorical nature. Each scaling method ensures that the data is appropriately transformed to improve model performance and stability.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, RobustScaler, MinMaxScaler, PowerTransformer, QuantileTransformer


# Initialize scalers
standard_scaler = StandardScaler()
robust_scaler = RobustScaler()
minmax_scaler = MinMaxScaler()
power_transformer = PowerTransformer(method='yeo-johnson')
quantile_transformer = QuantileTransformer(output_distribution='uniform')

# Apply scaling based on distributions
df["tempp"] = standard_scaler.fit_transform(df[["temp"]])  # StandardScaler for normally distributed data
df[["n", "p", "k"]] = robust_scaler.fit_transform(df[["n", "p", "k"]])  # RobustScaler for outliers
df["humidity"] = power_transformer.fit_transform(df[["hum"]])  # PowerTransformer for skewed data
# df["yield"] = power_transformer.fit_transform(df[["yield"]])  # PowerTransformer for yield (highly skewed)
df["ph"] = minmax_scaler.fit_transform(df[["ph"]])  # MinMaxScaler for pH (if continuous)

df_scaled = df
    
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Assuming you have the full dataset in 'df' with 'crop' as one of the columns
# Initialize empty lists for the training, testing, and validation sets
train_set = []
val_set = []
test_set = []

# List of unique crops in your dataset
crop_types = df['label'].unique()

# Iterate over each crop type and split its data
for crop in crop_types:
    crop_data = df[df['label'] == crop]
    
    # Split the crop data into training and temp (validation + test)
    train_data, temp_data = train_test_split(crop_data, test_size=0.3, random_state=42)
    
    # Split the temp data into validation and test sets
    val_data, test_data = train_test_split(temp_data, test_size=0.5, random_state=42)
    
    # Append the splits to the corresponding lists
    train_set.append(train_data)
    val_set.append(val_data)
    test_set.append(test_data)

# Concatenate the splits back into a single dataframe
train_df = pd.concat(train_set, axis=0)
val_df = pd.concat(val_set, axis=0)
test_df = pd.concat(test_set, axis=0)

# Shuffle the data (optional)
train_df = train_df.sample(frac=1, random_state=42).reset_index(drop=True)
val_df = val_df.sample(frac=1, random_state=42).reset_index(drop=True)
test_df = test_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Now you have separate train, validation, and test sets
```

```python
import gym
from gym import spaces
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

class FertilizerEnv(gym.Env):
    def __init__(self, df):
        super(FertilizerEnv, self).__init__()

        self.df = df
        self.crops = df['label'].unique()
        self.current_crop_idx = 0

        # Action space: N, P, K recommendations (scaled 0-1)
        self.action_space = spaces.Box(low=np.array([0, 0, 0]), high=np.array([1, 1, 1]), dtype=np.float32)

        # Observation space: Soil parameters (e.g., pH, temp, humidity, etc.)
        self.observation_space = spaces.Box(low=-1.0, high=1.0, shape=(6,), dtype=np.float32)

        # Standardize the dataset features
        self.scaler = StandardScaler()
        self.df_scaled = self.df.copy()
        self.df_scaled[['ph', 'temp', 'humidity', 'n', 'p', 'k']] = self.scaler.fit_transform(
            self.df[['ph', 'temp', 'humidity', 'n', 'p', 'k']]
        )

    def reset(self):
        """ Reset environment to the first crop and return its state """
        self.current_crop_idx = 0
        crop_data = self.df_scaled[self.df_scaled['label'] == self.crops[self.current_crop_idx]]

        # Ensure all 8 features are included correctly
        state = crop_data[['ph', 'temp', 'humidity', 'n', 'p', 'k']].iloc[0].values

        self.current_state = state
        return self.current_state


    def step(self, action):
        """ Take an action and return new state, reward, and done flag """
        n, p, k = action  # Fertilizer inputs

        # Get crop data
        crop_data = self.df_scaled[self.df_scaled['label'] == self.crops[self.current_crop_idx]]
        ideal_n, ideal_p, ideal_k = crop_data[['n', 'p', 'k']].iloc[0].values  

        # Reward function: minimize deviation from ideal NPK
        npk_difference = np.abs(np.array([n, p, k]) - np.array([ideal_n, ideal_p, ideal_k]))
        reward = -np.sum(npk_difference)

        # Move to the next crop (or reset)
        self.current_crop_idx += 1
        if self.current_crop_idx >= len(self.crops):
            self.current_crop_idx = 0

        # Get the next crop’s state
        crop_data = self.df_scaled[self.df_scaled['label'] == self.crops[self.current_crop_idx]]
        state = crop_data[['ph', 'temp', 'humidity', 'n', 'p', 'k']].iloc[0].values

        self.current_state = state
        done = False  
        return self.current_state, reward, done, {}


    def render(self):
        """ Optional: Display environment state """
        pass
```

```python
# Initialize the environment with your dataset
env = FertilizerEnv(df)

# Initialize the PPO model
model = PPO('MlpPolicy', env, verbose=1, tensorboard_log='./ppo_fertilizer/')

# Train the model
model.learn(total_timesteps=500000)


```

```python
# Evaluate the agent
state = env.reset()
total_reward = 0

for _ in range(100):  # Run for 100 steps
    action, _states = model.predict(state)
    state, reward, done, _ = env.step(action)
    total_reward += reward
    if done:
        break

print(f"Total Reward: {total_reward}")
```
