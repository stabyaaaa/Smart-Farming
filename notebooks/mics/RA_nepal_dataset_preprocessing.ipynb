```python
import pandas as pd
import ast  # Import for safely evaluating string literals as Python literals

# Define the crops and their varieties
crops_and_varieties = [
    ('Maize', 'OPV'),
    ('Maize', 'Hybrid'),
    ('Rice', 'OPV'),
    ('Rice', 'Hybrid'),
    ('Wheat', 'OPV')
]

# Prepare the new rows for each crop-variety combination with soil data
new_rows = []

# Extract the soil data from the first row (assuming all rows have the same soil data structure)
soil_data_columns = ['lat', 'lon', 'ph', 'organic_matter', 'total_nitrogen', 'potassium', 'p2o5',
                     'boron', 'zinc', 'sand', 'clay', 'slit', 'parentsoil', 'province', 'district', 'palika']

# Iterate over the dataset rows (all 39,000 rows)
for _, row in df.iterrows():
    # Extract soil data for this row
    soil_data = row[soil_data_columns].to_dict()

    # Extract fertilizer data, which may be a string, dictionary, NaN, or invalid type
    fertilizer_data = row['fertilizer']
    
    # Debugging: Log the type of the fertilizer_data
    print(f"Fertilizer data type for row {_}: {type(fertilizer_data)}")
    
    # Check if fertilizer_data is a valid string, dictionary, NaN or float
    if isinstance(fertilizer_data, str):
        try:
            # Safely evaluate string to dictionary if it's a valid string representation
            fertilizer_data = ast.literal_eval(fertilizer_data)
            print(f"Successfully converted fertilizer data for row {_}.")
        except Exception as e:
            print(f"❌ Error parsing fertilizer data for row {_}: {e}")
            continue  # Skip this row if parsing fails
    elif isinstance(fertilizer_data, dict):
        # If it's already a dictionary, use it directly
        print(f"Fertilizer data is already a dictionary for row {_}.")
    elif isinstance(fertilizer_data, float) or pd.isna(fertilizer_data):
        # If the fertilizer data is NaN or a float (invalid), log and skip this row
        print(f"❌ Invalid fertilizer data for row {_} (NaN or float), skipping this row.")
        continue
    else:
        # If fertilizer_data is of an unexpected type, log and skip it
        print(f"❌ Unexpected fertilizer data type in row {_}: {type(fertilizer_data)}")
        continue

    # Ensure fertilizer_data is a dictionary before continuing
    if not isinstance(fertilizer_data, dict):
        print(f"❌ Skipping row {_} because fertilizer data is not a dictionary.")
        continue

    # Now iterate over the crops and varieties to extract the fertilizer info
    for crop, variety in crops_and_varieties:
        try:
            # Check if crop and variety exist in the fertilizer_data
            if crop in fertilizer_data and variety in fertilizer_data[crop]:
                fert_row = fertilizer_data[crop][variety]

                # Create a new row with the soil data and fertilizer information
                new_row = soil_data.copy()
                new_row['crop'] = crop
                new_row['variety'] = variety
                new_row['UREA1'] = fert_row.get('UREA1', 'NA')  # Default 'NA' if missing
                new_row['UREA2'] = fert_row.get('UREA2', 'NA')  # Default 'NA' if missing
                new_row['UREA3'] = fert_row.get('UREA3', 'NA')  # Default 'NA' if missing
                new_row['DAP'] = fert_row.get('DAP', 'NA')      # Default 'NA' if missing
                new_row['MOP'] = fert_row.get('MOP', 'NA')      # Default 'NA' if missing
                new_row['organic'] = fert_row.get('organic', 'NA')  # Default 'NA' if missing
                new_row['boron_fert'] = fert_row.get('boron', 'NA')  # Default 'NA' if missing
                
                # Optionally add zinc_fert (skip if not necessary)
                # new_row['zinc_fert'] = fert_row.get('zinc', 'NA')  # You can set default 'NA' if missing

                # Append the new row to the list
                new_rows.append(new_row)
        except KeyError as e:
            print(f"❌ Missing expected crop/variety in fertilizer data for row {_}: {e}")
            continue  # Skip this row if there's a KeyError

# Convert the list of new rows into a DataFrame
dff = pd.DataFrame(new_rows)

# Optionally write the expanded dataset to a new CSV file
dff.to_csv('expanded_fertilizer_recommendations.csv', index=False)

print("Dataset has been adjusted and saved as 'expanded_fertilizer_recommendations.csv'.")
```
