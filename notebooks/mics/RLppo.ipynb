```python
import torch
# Check if CUDA is available
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using {device} device")
```

```python
class FertilizerEnv:
    def __init__(self, X, y):
        self.X = X.values.astype(np.float32)
        self.y = y.values.astype(np.float32)
        self.index = 0

    def reset(self):
        self.index = 0
        return self.X[self.index]

    def step(self, action):
        true = self.y[self.index]
        mse = np.mean((true - action)**2)
        reward = -mse / 10  # scaled reward
        self.index += 1
        done = self.index >= len(self.X)
        next_state = self.X[self.index] if not done else np.zeros_like(self.X[0])
        return next_state, reward, done
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.utils import shuffle

import time
# Device setup
# device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
subset_size = 128  # or any smaller value
indices = np.random.choice(len(X_train), subset_size, replace=False)
X_sample = X_train.iloc[indices]
y_sample = y_train.iloc[indices]
env = FertilizerEnv(X_sample, y_sample)

# Simulated Environment for PPO Training


# Actor-Critic Network
class ActorCritic(nn.Module):
    def __init__(self, state_dim, action_dim):
        super(ActorCritic, self).__init__()
        self.actor = nn.Sequential(
            nn.Linear(state_dim, 64),
            nn.ReLU(),
            nn.Linear(64, action_dim),
            nn.Tanh()  # output bounded between -1 and 1
        )
        self.critic = nn.Sequential(
            nn.Linear(state_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 1)
        )
    
    def forward(self, state):
        return self.actor(state), self.critic(state)
    
    def select_action(self, state):
        with torch.no_grad():
            action, _ = self.forward(state)
            return action * 3  # Rescale output to -3 to +3


# PPO Agent
class PPOAgent:
    def __init__(self, state_dim, action_dim, lr=1e-3, gamma=0.99, update_epochs=2, batch_size=64):
        self.gamma = gamma
        self.update_epochs = update_epochs
        self.batch_size = batch_size
        self.model = ActorCritic(state_dim, action_dim).to(device)
        self.optimizer = optim.Adam(self.model.parameters(), lr=lr)
        self.mse_loss = nn.MSELoss()

    def get_action(self, state):
        state = torch.tensor(state, dtype=torch.float32).to(device)
        with torch.no_grad():
            action, value = self.model(state)
        return (action * 3).cpu().numpy(), value.item()

    def update(self, states, actions, rewards, values):
        states = torch.tensor(states, dtype=torch.float32).to(device)
        actions = torch.tensor(actions, dtype=torch.float32).to(device)
        values = torch.tensor(values, dtype=torch.float32).to(device)

        # Compute returns
        returns = []
        G = 0
        for r in reversed(rewards):
            G = r + self.gamma * G
            returns.insert(0, G)
        returns = torch.tensor(returns, dtype=torch.float32).to(device)

        advantages = returns - values

        dataset = list(zip(states, actions, returns, advantages))
        for _ in range(self.update_epochs):
            shuffled = shuffle(dataset)
            for i in range(0, len(shuffled), self.batch_size):
                batch = shuffled[i:i+self.batch_size]
                if not batch:
                    continue
                b_states, b_actions, b_returns, b_advantages = zip(*batch)
                b_states = torch.stack(b_states)
                b_actions = torch.stack(b_actions)
                b_returns = torch.stack(b_returns)
                b_advantages = torch.stack(b_advantages)

                new_actions, new_values = self.model(b_states)
                action_loss = self.mse_loss(new_actions, b_actions)
                value_loss = self.mse_loss(new_values.squeeze(), b_returns)
                loss = action_loss + 0.5 * value_loss

                self.optimizer.zero_grad()
                loss.backward()
                self.optimizer.step()


# === Training Loop ===
def train_ppo(X_train, y_train, epochs=10):
    env = FertilizerEnv(X_train, y_train)
    state_dim = X_train.shape[1]
    action_dim = y_train.shape[1]
    agent = PPOAgent(state_dim, action_dim)

    for epoch in range(epochs):
        start_time = time.time()
        state = env.reset()
        states, actions, rewards, values = [], [], [], []
        total_reward = 0

        while True:
            action, value = agent.get_action(state)
            next_state, reward, done = env.step(action)

            total_reward += reward
            states.append(state)
            actions.append(action)
            rewards.append(reward)
            values.append(value)

            state = next_state
            if done:
                break

        agent.update(states, actions, rewards, values)

        print(f"Epoch {epoch+1}/{epochs} | Total Reward: {total_reward:.4f} | Time: {time.time() - start_time:.2f}s")

    # Evaluation
    with torch.no_grad():
        X_tensor = torch.tensor(X_train.values, dtype=torch.float32).to(device)
        predictions, _ = agent.model(X_tensor)
        predictions = (predictions * 3).cpu().numpy()

    mse = mean_squared_error(y_train, predictions)
    mae = mean_absolute_error(y_train, predictions)
    r2 = r2_score(y_train, predictions)
    print(f"\n🔍 Final Metrics - MSE: {mse:.4f}, MAE: {mae:.4f}, R²: {r2:.4f}")
```
