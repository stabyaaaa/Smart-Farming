```python
# ==========================================================
# PPO FROM SCRATCH FOR FERTILIZER RECOMMENDATION (NARC)
# Saves BEST model based on LOWEST VALIDATION RMSE
# Targets: UREA, DAP, MOP (kg/ropani)
# ==========================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.distributions import Normal
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import mean_absolute_error, r2_score
import random

# ==========================================================
# REPRODUCIBILITY & DEVICE
# ==========================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

EPS = 1e-8

# ==========================================================
# LOAD DATA
# ==========================================================

fert_cols = ["UREA1", "DAP", "MOP"]

# Encode categorical variables
for col in ["crop", "variety", "parentsoil"]:
    if col in df.columns:
        df[col] = LabelEncoder().fit_transform(df[col].astype(str))

# Manually select features you want to use for training
feature_cols = [
    'ph',
    'organic_matter',
    'total_nitrogen',
    'p2o5',
    'potassium',
    'boron',
    'zinc',
    'sand',
    'clay',
    'slit',
    'boron'
]

# Optional: if your dataset has extra columns, make sure to only pick existing ones
feature_cols = [col for col in feature_cols if col in df.columns]
X = df[feature_cols].values
y = df[fert_cols].values   # real units (kg/ropani)
```

```python


# ==========================================================
# SCALE INPUTS ONLY
# ==========================================================
scaler_X = StandardScaler()
X = scaler_X.fit_transform(X)

X = torch.tensor(X, dtype=torch.float32, device=device)
y = torch.tensor(y, dtype=torch.float32, device=device)

# ==========================================================
# TRAIN / VALIDATION / TEST SPLIT
# ==========================================================
perm = torch.randperm(len(X))
n = len(X)
train_end = int(0.7 * n)
val_end = int(0.85 * n)

X_train, y_train = X[perm[:train_end]], y[perm[:train_end]]
X_val, y_val     = X[perm[train_end:val_end]], y[perm[train_end:val_end]]
X_test, y_test   = X[perm[val_end:]], y[perm[val_end:]]

state_dim = X_train.shape[1]
action_dim = 3

# ==========================================================
# ACTION BOUNDS (kg/ropani)
# ==========================================================
FERT_MIN = torch.tensor([0.0, 0.5, 0.4], device=device)
FERT_MAX = torch.tensor([3.3, 4.4, 3.4], device=device)

# ==========================================================
# PPO HYPERPARAMETERS (BEST CONFIG)
# ==========================================================
hidden_dim = 256
lr_actor = 1e-4
lr_critic = 5e-4
clip_eps = 0.2
ppo_epochs = 10
steps_per_update = 2048
mini_batch = 64
entropy_coef = 0.005
max_grad_norm = 0.5
num_updates = 1500

# ==========================================================
# ACTOR & CRITIC
# ==========================================================
class Actor(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
        )
        self.mu = nn.Linear(hidden_dim, action_dim)
        self.log_std = nn.Parameter(torch.ones(action_dim) * -0.5)

    def forward(self, x):
        h = self.net(x)
        mu = self.mu(h)
        std = torch.exp(self.log_std)
        return mu, std

class Critic(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 1),
        )

    def forward(self, x):
        return self.net(x)

actor = Actor().to(device)
critic = Critic().to(device)

opt_actor = optim.Adam(actor.parameters(), lr=lr_actor)
opt_critic = optim.Adam(critic.parameters(), lr=lr_critic)

# ==========================================================
# REWARD FUNCTION
# ==========================================================
def reward_fn(pred, truth):
    return -torch.abs(pred - truth).sum(dim=1, keepdim=True)

# ==========================================================
# RMSE EVALUATION (PER ACTION)
# ==========================================================
@torch.no_grad()
def eval_rmse(X, y):
    mu, _ = actor(X)
    a = torch.tanh(mu)
    a_real = FERT_MIN + (a + 1) / 2 * (FERT_MAX - FERT_MIN)
    rmse = torch.sqrt(((a_real - y) ** 2).mean(dim=0))
    return rmse

# ==========================================================
# CHECKPOINT SETUP
# ==========================================================
best_val_rmse = float("inf")
best_update = -1
save_dir = "final_ppo_model"
os.makedirs(save_dir, exist_ok=True)
save_path = os.path.join(save_dir, "ppo_fertnet_best.pt")

# ==========================================================
# PPO TRAINING LOOP
# ==========================================================
for update in range(1, num_updates + 1):

    S, A, R, V, LP = [], [], [], [], []

    # ------------------ ROLLOUT ------------------
    while len(S) * mini_batch < steps_per_update:
        idx = torch.randint(0, len(X_train), (mini_batch,))
        s = X_train[idx]
        t = y_train[idx]

        with torch.no_grad():
            mu, std = actor(s)
            dist = Normal(mu, std)

            raw_a = dist.rsample()
            a = torch.tanh(raw_a)

            a_real = FERT_MIN + (a + 1) / 2 * (FERT_MAX - FERT_MIN)

            logp = dist.log_prob(raw_a).sum(-1, keepdim=True)
            logp -= torch.sum(torch.log(1 - a**2 + EPS), dim=1, keepdim=True)

            v = critic(s)
            r = reward_fn(a_real, t)

        S.append(s); A.append(raw_a)
        R.append(r); V.append(v); LP.append(logp)

    S = torch.cat(S)
    A = torch.cat(A)
    R = torch.cat(R)
    V = torch.cat(V)
    LP = torch.cat(LP)

    # ------------------ ADVANTAGE ------------------
    adv = R - V
    adv = (adv - adv.mean()) / (adv.std() + EPS)

    # ------------------ PPO UPDATE ------------------
    for _ in range(ppo_epochs):
        idx = torch.randperm(len(S))
        for i in range(0, len(S), mini_batch):
            mb = idx[i:i+mini_batch]

            mu, std = actor(S[mb])
            dist = Normal(mu, std)
            entropy = dist.entropy().sum(dim=1).mean()

            new_lp = dist.log_prob(A[mb]).sum(-1, keepdim=True)
            ratio = torch.exp(new_lp - LP[mb])

            surr1 = ratio * adv[mb]
            surr2 = torch.clamp(ratio, 1-clip_eps, 1+clip_eps) * adv[mb]

            actor_loss = -torch.min(surr1, surr2).mean() - entropy_coef * entropy
            critic_loss = nn.MSELoss()(critic(S[mb]), R[mb])

            opt_actor.zero_grad()
            opt_critic.zero_grad()
            actor_loss.backward()
            critic_loss.backward()

            torch.nn.utils.clip_grad_norm_(actor.parameters(), max_grad_norm)
            torch.nn.utils.clip_grad_norm_(critic.parameters(), max_grad_norm)

            opt_actor.step()
            opt_critic.step()

    # ------------------ EVALUATION ------------------
    train_rmse = eval_rmse(X_train, y_train)
    val_rmse   = eval_rmse(X_val, y_val)
    mean_val_rmse = val_rmse.mean().item()

    if mean_val_rmse < best_val_rmse:
        best_val_rmse = mean_val_rmse
        best_update = update

        torch.save({
            "actor_state_dict": actor.state_dict(),
            "critic_state_dict": critic.state_dict(),
            "scaler_X": scaler_X,
            "feature_cols": feature_cols,
            "fert_cols": fert_cols,
            "best_update": update,
            "best_val_rmse": best_val_rmse
        }, save_path)

    if update % 10 == 0:
        print(
            f"[{update:04d}] "
            f"Train RMSE {train_rmse.cpu().numpy()} | "
            f"Val RMSE {val_rmse.cpu().numpy()} | "
            f"BEST Val RMSE {best_val_rmse:.4f} @ update {best_update}"
        )

# ==========================================================
# FINAL EVALUATION ON TEST SET
# ==========================================================
@torch.no_grad()
def eval_metrics(X, y):
    mu, _ = actor(X)
    a = torch.tanh(mu)
    a_real = FERT_MIN + (a + 1) / 2 * (FERT_MAX - FERT_MIN)
    rmse = torch.sqrt(((a_real - y) ** 2).mean(dim=0))
    mae  = torch.tensor([mean_absolute_error(y[:,i].cpu(), a_real[:,i].cpu()) for i in range(y.shape[1])])
    r2   = torch.tensor([r2_score(y[:,i].cpu(), a_real[:,i].cpu()) for i in range(y.shape[1])])
    return rmse, mae, r2

# Load best model
checkpoint = torch.load(save_path)
actor.load_state_dict(checkpoint["actor_state_dict"])
critic.load_state_dict(checkpoint["critic_state_dict"])

test_rmse, test_mae, test_r2 = eval_metrics(X_test, y_test)

print("\n✅ FINAL TEST SET METRICS (PER FERTILIZER)")
for i, fert in enumerate(fert_cols):
    print(f"{fert} - RMSE: {test_rmse[i]:.4f}, MAE: {test_mae[i]:.4f}, R2: {test_r2[i]:.4f}")

print(f"\nBest model saved at update {best_update} with Val RMSE: {best_val_rmse:.4f}")
print("Saved to:", save_path)
```
